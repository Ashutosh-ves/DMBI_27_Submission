# Experiment 3 — Exploratory Data Analysis and Visualization

### Step 1: Loading the Dataset and Initial Exploration

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Load the original (raw, uncleaned) dataset — same file used in Experiment 2
df = pd.read_csv('application_matching_dataset.csv')

print("Dataset Info:")
print(df.info())
print("\nFirst 5 Rows:")
print(df.head())
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title("Missing Value Pattern Heatmap")
plt.xlabel("Columns")
plt.show()

print((df.isnull().mean() * 100).round(1))
```

### Step 2: Task 1 — Descriptive Analysis: Statistical Measures of Central Tendency

```python
numeric_cols = ['cgpa', 'match_score', 'time_to_response_days', 'skill_gap_count']
categorical_cols = ['branch', 'city', 'opening_type', 'industry', 'status_label']

central_tendency = {
    'Mean': df[numeric_cols].mean(),
    'Median': df[numeric_cols].median(),
    'Mode': df[categorical_cols].mode().iloc[0]
}

print("Central Tendency Measures:")
for measure, values in central_tendency.items():
    print(f"\n{measure}:")
    print(values)
```

### Step 3: Task 2 — Descriptive Analysis: Statistical Measures of Dispersion

```python
dispersion = {
    'Standard Deviation': df[numeric_cols].std(),
    'Variance': df[numeric_cols].var(),
    'Range': df[numeric_cols].max() - df[numeric_cols].min(),
    'IQR': df[numeric_cols].quantile(0.75) - df[numeric_cols].quantile(0.25)
}

print("Dispersion Measures:")
for measure, values in dispersion.items():
    print(f"\n{measure}:")
    print(values)
```

### Step 4: Task 3 — Correlation Between Attributes

```python
# Create a numeric version of the outcome for correlation purposes
df['is_selected'] = (df['status_label'] == 'selected').astype(int)

correlation_matrix = df[['cgpa', 'match_score', 'time_to_response_days',
                          'skill_gap_count', 'is_selected']].corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

### Step 4b: Task 3b — Groupby Aggregation Analysis

```python
avg_match_by_industry = df.groupby('industry')['match_score'].mean().sort_values(ascending=False)
print("Average match_score by industry:")
print(avg_match_by_industry)

avg_match_by_branch = df.groupby('branch')['match_score'].mean().sort_values(ascending=False)
print("\nAverage match_score by branch (raw, uncleaned text):")
print(avg_match_by_branch)
print("\nNumber of distinct branch labels found:", df['branch'].nunique())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# 1. Histogram for match_score
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="match_score", bins=30, kde=True)
plt.title("Histogram of Match Score")
plt.xlabel("Match Score")
plt.ylabel("Count")
plt.show()

# 2. Boxplot for cgpa by status_label
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="status_label", y="cgpa")
plt.title("Boxplot of CGPA by Status Label")
plt.xlabel("Status Label")
plt.ylabel("CGPA")
plt.show()

# 3. Heatmap of Correlation Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()

# 4. Scatter Plot for cgpa vs match_score
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="cgpa", y="match_score", hue="status_label")
plt.title("Scatter Plot of CGPA vs Match Score")
plt.xlabel("CGPA")
plt.ylabel("Match Score")
plt.show()

# 5. Bar Chart — count of applications by status
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="status_label", order=df['status_label'].value_counts().index)
plt.title("Count of Applications by Status Label")
plt.xlabel("Status Label")
plt.ylabel("Count")
plt.show()

# 6. Bar Chart — count of applications by industry
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y="industry", order=df['industry'].value_counts().index)
plt.title("Count of Applications by Industry")
plt.xlabel("Count")
plt.ylabel("Industry")
plt.show()
```

```python
stipend_numeric = pd.to_numeric(df['expected_stipend'], errors='coerce')

plt.figure(figsize=(10, 6))
sns.histplot(stipend_numeric.dropna(), bins=30, kde=True)
plt.title("Histogram of Expected Stipend (raw, uncleaned)")
plt.xlabel("Expected Stipend")
plt.show()

print("Skewness of expected_stipend:", stipend_numeric.skew())
```
