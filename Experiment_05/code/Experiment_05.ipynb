# Experiment 5 — Classification using Python and User-Defined Functions

### Step 1: Load the Dataset and Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

df = pd.read_csv('application_matching_exp5.csv')
print(df.info())
print(df.head())
```

### Step 2: Data Preprocessing

```python
# Select Features (X) and Target (y)
features = ['cgpa', 'match_score', 'time_to_response_days',
            'skill_gap_count', 'expected_stipend', 'year_of_study']
X = df[features]
y = df['Elite_Match']

# Normalize data using StandardScaler — these features are on very different
# scales (cgpa ~5-10, expected_stipend in the thousands)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split Data (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
```

### Step 3: Train a Decision Tree Model (Inbuilt, Gini)

```python
# ---- Step 3: Train a Decision Tree Model ----
dt_model = DecisionTreeClassifier(criterion="gini", max_depth=5, random_state=42)
dt_model.fit(X_train, y_train)
```

### Step 4: Evaluate Model Performance (Inbuilt Decision Tree)

```python
y_pred_dt = dt_model.predict(X_test)

print("Decision Tree Model Accuracy:", accuracy_score(y_test, y_pred_dt))
print("\nClassification Report:\n", classification_report(y_test, y_pred_dt))

plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, y_pred_dt), annot=True, fmt='d', cmap="Blues",
            xticklabels=["Not-Elite", "Elite"], yticklabels=["Not-Elite", "Elite"])
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("Decision Tree (Gini) - Confusion Matrix")
plt.show()
```

### Step 5: Build a Decision Tree Using a Custom (Student-Defined) Function

```python
# ---- Step 4: Build a Decision Tree Model using a Custom Function ----
def custom_decision_tree(X_train, y_train, X_test):
    # Using 'entropy' instead of 'gini' as the splitting criterion
    model = DecisionTreeClassifier(criterion="entropy", max_depth=5, random_state=42)
    model.fit(X_train, y_train)
    return model.predict(X_test)

y_pred_custom_dt = custom_decision_tree(X_train, y_train, X_test)

print("Custom Decision Tree Model Accuracy:", accuracy_score(y_test, y_pred_custom_dt))
print("\nCustom Model Classification Report:\n", classification_report(y_test, y_pred_custom_dt))
```

### Step 6: Visualize the Decision Tree (Text-Based)

```python
tree_rules = export_text(dt_model, feature_names=features, max_depth=3)
print(tree_rules)
```

### Step 1: Train a Naïve Bayes Model (Inbuilt)

```python
from sklearn.naive_bayes import GaussianNB

# ---- Train a Naïve Bayes Model ----
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
```

### Step 2: Evaluate the Naïve Bayes Model

```python
y_pred_nb = nb_model.predict(X_test)

print("Naive Bayes Model Accuracy:", accuracy_score(y_test, y_pred_nb))
print("\nClassification Report:\n", classification_report(y_test, y_pred_nb))

plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, y_pred_nb), annot=True, fmt='d', cmap="Blues",
            xticklabels=["Not-Elite", "Elite"], yticklabels=["Not-Elite", "Elite"])
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("Naive Bayes - Confusion Matrix")
plt.show()
```

### Step 3: Build a Naïve Bayes Model Using a Custom Function

```python
def custom_naive_bayes(X_train, y_train, X_test):
    model = GaussianNB()
    model.fit(X_train, y_train)
    return model.predict(X_test)

y_pred_custom_nb = custom_naive_bayes(X_train, y_train, X_test)

print("Custom Naive Bayes Model Accuracy:", accuracy_score(y_test, y_pred_custom_nb))
print("\nCustom Model Classification Report:\n", classification_report(y_test, y_pred_custom_nb))
```
